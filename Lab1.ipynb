# BÀI 1: LÀM QUEN VỚI PYTORCH

### Bài 1. Tạo Tensor cơ bản

```python
import torch
a = torch.tensor(
    [[1, 2, 3],
     [4, 5, 6]],
    dtype=torch.int32
)
print(a)
```

## Giải thích

- **Kích thước:** `torch.Size([2, 3])`, tensor có 2 hàng và 3 cột.
- **Số chiều:** `2`, vì tensor có 2 chiều.
- **Kiểu dữ liệu:** `torch.int32`, các phần tử là số nguyên 32-bit.

### Bài 2. Tạo các Tensor thường dùng

```python
import torch
a = torch.rand(3, 4)
b = torch.randn(3, 4)
c = torch.zeros(2, 3)
d = torch.ones(2, 3)
e = torch.eye(4)

print(a)
print(b)
print(c)
print(d)
print(e)
```

## Giải thích

- `torch.rand()` tạo tensor có các giá trị ngẫu nhiên theo phân phối đều trong khoảng `[0, 1)`.
- `torch.randn()` tạo tensor có các giá trị ngẫu nhiên theo phân phối chuẩn, với trung bình gần `0` và độ lệch chuẩn gần `1`.
- Tensor `f` có kích thước `5 × 5` và tất cả phần tử đều bằng `10`, được tạo bằng `torch.full((5, 5), 10)`.

### Bài 3. Kiểm tra thông tin Tensor

```python
import torch
a = torch.zeros(2, 3, 4)
print("Tensor:")
print(a)
print("Shape:", a.shape)
print("Size:", a.size())
print("Number of dimensions:", a.ndim)
print("Data type:", a.dtype)
```

## Giải thích

`torch.Size([2, 3, 4])` cho biết tensor `a` có **3 chiều**:

- `2`: có 2 khối dữ liệu.
- `3`: mỗi khối có 3 hàng.
- `4`: mỗi hàng có 4 phần tử.

Vì vậy, tensor có dạng `2 × 3 × 4` và chứa tổng cộng `24` phần tử.

### Bài 4. Slicing và Indexing

```python
import torch
a = torch.randn(3, 4, 5)
print("Original tensor:")
print(a)
print("\nTensor a[1]:")
print(a[1])
print("\nTensor a[1:, 2:4]:")
print(a[1:, 2:4])
```

## Giải thích

- `a[1]` lấy phần tử tại vị trí `1` trên chiều đầu tiên, nên có shape `torch.Size([4, 5])`.
- `a[1:, 2:4]` lấy từ khối thứ hai trở đi, lấy hàng ở vị trí `2` và `3`, đồng thời giữ toàn bộ chiều cuối; shape là `torch.Size([2, 2, 5])`.
- Để lấy hai phần tử cuối trên chiều thứ ba, dùng `a[:, :, -2:]`.

### Bài 5. Reshape và Flatten

```python
import torch
a = torch.randn(3, 4)
print("Original:")
print(a)
print("\nFlatten:")
print(a.ravel())
print("\nReshape to 3 x 2 x 2:")
print(a.reshape(3, 2, 2))
```

## Giải thích

- Tensor ban đầu có `3 × 4 = 12` phần tử.
- `a.ravel()` làm phẳng tensor thành một chiều nhưng vẫn giữ nguyên `12` phần tử.
- `a.reshape(3, 2, 2)` đổi hình dạng thành `3 × 2 × 2`, vẫn có `12` phần tử.

### Bài 6. Phép toán và nhân ma trận

```python
import torch
a = torch.randn(2, 3)
b = torch.randn(2, 3)
print("a:")
print(a)
print("b:")
print(b)
print("\na + b:")
print(a + b)
print("\na / b:")
print(a / b)
print("\na^2:")
print(a ** 2)
print("\nMatrix multiplication:")
print(a @ b.T)
```

## Giải thích

- `a * b` là phép nhân từng phần tử tương ứng; nếu thực hiện thì kết quả có shape `torch.Size([2, 3])`.
- `a @ b.T` là phép nhân ma trận: `a` có shape `(2, 3)`, `b.T` có shape `(3, 2)`, nên kết quả có shape `torch.Size([2, 2])`.
- Hai phép toán này khác nhau về cách kết hợp các phần tử: một phép nhân từng vị trí, một phép nhân theo hàng và cột.

### Bài 7. Các hàm thống kê trên Tensor

```python
import torch
a = torch.randn(3, 4)
print(a)
print("\nMean:")
print(torch.mean(a, dim=0))
print("\nStandard deviation:")
print(torch.std(a, dim=0))
print("\nCumulative sum:")
print(torch.cumsum(a, dim=0))
```

## Giải thích

- Với `dim=0`, phép tính được thực hiện theo chiều đầu tiên, nên tensor `(3, 4)` cho kết quả có shape `(4,)`, tương ứng với từng cột.
- `torch.mean` tính giá trị trung bình, `torch.std` tính độ lệch chuẩn, còn `torch.cumsum` tính tổng tích lũy theo chiều được chọn.
- Nếu thay `dim=0` bằng `dim=1`, các phép tính sẽ thực hiện theo từng hàng và kết quả có shape `(3,)`.

### Bài 8. Autograd cơ bản

```python
import torch
x = torch.tensor(
    3.6,
    requires_grad=True
)
y = x * x
y.backward()
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)
```

## Giải thích

- Code đặt `x = 3.6` với `requires_grad=True`, nên PyTorch theo dõi phép tính để tính đạo hàm.
- Với `y = x²`, đạo hàm là `dy/dx = 2x`. Tại `x = 3.6`, giá trị thủ công là `7.2`, khớp với `x.grad`.
- `y.backward()` thực hiện lan truyền ngược để tính gradient.

### Bài 9. Gradient Descent tìm hệ số đa thức

```python
import numpy as np
import torch
polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:")
print(w)
for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()
print("Learned coefficients:")
print(w)
```

## Giải thích

- Ba hệ số trong `w` lần lượt là hệ số của `x²`, `x` và hằng số, nên được so sánh với `[1, 2, 3]`.
- `optimizer.zero_grad()` xóa gradient của vòng lặp trước.
- `mse.backward()` tính gradient của hàm mất mát `mse` theo `w`.
- `optimizer.step()` cập nhật các hệ số `w` để giảm sai số.

### Bài 10. Giải hệ phương trình bằng Autograd

```python
import random
import torch
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)
EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)
for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
optimizer.zero_grad()
sqerr.backward()
optimizer.step()
print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)
```

## Giải thích

- `y1`, `y2`, `y3` và `y4` là độ lệch của bốn phương trình so với vế phải tương ứng.
- Khi các residual gần `0`, các giá trị `A`, `B`, `C`, `D` gần thỏa mãn: `A + B = 9`, `C - D = 1`, `A + C = 8` và `B - D = 2`.
- Bốn phương trình không độc lập hoàn toàn nên hệ có thể có nhiều nghiệm. Đặt `A = t` thì một họ nghiệm là `B = 9 - t`, `C = 8 - t`, `D = 7 - t`.

### Bài 11. Tổng hợp các thao tác với Tensor

```python
import numpy as np
import torch

# Tạo Tensor X có 24 phần tử và reshape theo ba kích thước
X = torch.arange(24)
print("X ban đầu:", X)
print("X reshape 4 x 6:")
print(X.reshape(4, 6))
print("X reshape 2 x 12:")
print(X.reshape(2, 12))
print("X reshape 2 x 3 x 4:")
print(X.reshape(2, 3, 4))

# Các phép toán trên hai Tensor A và B
A = torch.randn(3, 4)
B = torch.randn(3, 4)
print("\nA + B:")
print(A + B)
print("A * B:")
print(A * B)
print("A / B:")
print(A / B)
print("Mean theo cột của A:")
print(torch.mean(A, dim=0))
print("Độ lệch chuẩn theo cột của A:")
print(torch.std(A, dim=0))

# Tính đạo hàm bằng autograd
x_autograd = torch.tensor(2.0, requires_grad=True)
y_autograd = 2 * x_autograd ** 2 + 5 * x_autograd + 3
y_autograd.backward()
print("\n dy/dx bằng autograd:", x_autograd.grad.item())
print(" dy/dx thủ công:", 4 * x_autograd.item() + 5)

# Huấn luyện đa thức y = 2x^2 - 3x + 5
polynomial_11 = np.poly1d([2, -3, 5])
N_11 = 20
X_11 = np.random.randn(N_11, 1) * 5
Y_11 = polynomial_11(X_11)
XX_11 = np.hstack([
    X_11 * X_11,
    X_11,
    np.ones_like(X_11)
])
w_11 = torch.randn(3, 1, requires_grad=True)
x_11 = torch.tensor(XX_11, dtype=torch.float32)
y_11 = torch.tensor(Y_11, dtype=torch.float32)
optimizer_11 = torch.optim.NAdam([w_11], lr=0.01)
for _ in range(5000):
    optimizer_11.zero_grad()
    y_pred_11 = x_11 @ w_11
    mse_11 = torch.mean(torch.square(y_11 - y_pred_11))
    mse_11.backward()
    optimizer_11.step()
print("\nHệ số học được:", w_11.detach().flatten())
print("Hệ số mong đợi:", torch.tensor([2.0, -3.0, 5.0]))
```

## Giải thích

- `X` có `24` phần tử nên có thể reshape thành `4 × 6`, `2 × 12` hoặc `2 × 3 × 4`; tổng số phần tử vẫn giữ nguyên.
- `A + B`, `A * B` và `A / B` đều là phép toán từng phần tử trên hai tensor cùng shape `3 × 4`. Trung bình và độ lệch chuẩn theo cột dùng `dim=0`, nên mỗi kết quả có `4` giá trị.
- Với `y = 2x² + 5x + 3`, ta có `dy/dx = 4x + 5`. Tại `x = 2`, đạo hàm bằng `13`, khớp với kết quả của autograd.
- Đa thức huấn luyện có dạng `2x² - 3x + 5`, vì vậy ba hệ số mong đợi lần lượt là `[2, -3, 5]`. Các hệ số học được sẽ tiến gần các giá trị này.
